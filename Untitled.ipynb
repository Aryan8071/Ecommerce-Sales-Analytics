```python
delivered_orders=orders[orders['status']=="Delivered"]
```

```python
delivered_orders
```

```python
delivered_sales=order_items.merge(delivered_orders[["order_id"]],on='order_id',how='inner')
```

```python
delivered_sales
```

```python
delivered_revene=delivered_sales['line_total'].sum()
delivered_revene
```

```python
aov=delivered_revene/delivered_orders['order_id'].nunique()
aov
```

```python
return_rate=(orders['status'].eq("Returned").sum()/len(orders))*100
return_rate
```

```python
cancel_rate=(orders["status"].eq("Cancelled").sum()/len(orders))*100
cancel_rate
```

```python
delivery_rate=(orders["status"].eq("Delivered").sum()/len(orders))*100
delivery_rate
```

```python
orders["month"]=orders['order_date'].dt.to_period("M")
```

```python
monthly_order=orders.groupby('month')['order_id'].nunique()
monthly_order
```

```python
monthly_revenue=delivered_sales.copy()

monthly_revenue['month']=monthly_revenue['order_id'].map(orders.set_index('order_id')['month'])
monthly_revenue=monthly_revenue.groupby('month')['line_total'].sum()

monthly_revenue
```

```python
np.argmin(monthly_revenue)
```

```python
product_revenue=order_items.groupby('product_id')['line_total'].sum().sort_values(ascending=False)
product_revenue.head(10)
```

```python
products[products['product_id']=='P0119']
```

```python
category_revenue=order_items.merge(products[['product_id','category']],on='product_id',how='left').groupby('category')['line_total'].sum().sort_values (ascending=False)
category_revenue                                                   
```

```python
customers_revene=orders.merge(
    customers[['customer_id']],
    on='customer_id',
    how='left'
).merge(
    delivered_sales[['order_id','line_total']],
    on='order_id',
    how='inner'
).groupby('customer_id')['line_total'].sum().sort_values(ascending=False)

customers_revene.head(10)
```

```python
category_revenue=category_revenue.reset_index()
category_revenue.head(10)
```
